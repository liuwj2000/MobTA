```python
import pandas as pd
```

```python
bus=pd.read_csv('../data/SG/bus_with_grid_id_new.csv')
bus
```

```python
bus = bus[bus['arrive_time'] != "-"]
bus
```

```python
bus['min_of_day'] = pd.to_datetime(bus['arrive_time'], format='%H:%M').dt.hour * 60 + pd.to_datetime(bus['arrive_time'], format='%H:%M').dt.minute
bus = bus.sort_values(by=['trajectory_id', 'arrive_time']).reset_index(drop=True)
bus
```

```python
bus['prev_id'] = bus['trajectory_id'].shift(1)
bus['prev_min'] = bus['min_of_day'].shift(1)
bus['diff'] = bus['min_of_day'] - bus['prev_min']

# 找出“脏ID”：同一个ID内部，出现了 > 60 分钟间隔
# 这里的逻辑是：只要这条轨迹里有哪怕一次瞬移，整条轨迹我都不要了，以防万一
dirty_mask = (bus['trajectory_id'] == bus['prev_id']) & (bus['diff'] > 30)
dirty_ids = bus.loc[dirty_mask, 'trajectory_id'].unique()

print(f"发现脏轨迹(含大断层)数量: {len(dirty_ids)}")
```

```python
bus = bus[~bus['trajectory_id'].isin(dirty_ids)].copy()
bus
```

```python
import numpy as np
def split_long_trajectories(df, threshold=60):
    """
    对长度超过 threshold 的轨迹进行中点切分。
    """
    # 1. 首先确保轨迹内是有序的（按时间排序）
    df = df.sort_values(['trajectory_id', 'arrive_time'])
    
    # 2. 计算每个轨迹的长度
    traj_lengths = df.groupby('trajectory_id').size()
    
    # 3. 区分需要切分和不需要切分的 ID
    long_traj_ids = traj_lengths[traj_lengths > threshold].index
    print(f"共有 {len(long_traj_ids)} 条轨迹长度超过 {threshold}，即将进行切分。")

    # 4. 定义切分函数（用于 apply）
    def process_group(group):
        n = len(group)
        if n > threshold:
            mid = n // 2
            # 第一部分
            part1 = group.iloc[:mid].copy()
            part1['trajectory_id'] = part1['trajectory_id'].astype(str) + '_part1'
            # 第二部分
            part2 = group.iloc[mid:].copy()
            part2['trajectory_id'] = part2['trajectory_id'].astype(str) + '_part2'
            return pd.concat([part1, part2])
        else:
            return group

        # 5. 执行切分操作
    # 使用 groupby().apply() 能够保留所有原始列
    df_processed = df.groupby('trajectory_id', group_keys=False).apply(process_group)
    '''
    apply 执行完毕后，Pandas 会收集所有函数返回的 DataFrame，并将它们纵向堆叠（Stack）起来
    '''
    
    return df_processed

bus_new = split_long_trajectories(bus, threshold=60)
bus_new
```
