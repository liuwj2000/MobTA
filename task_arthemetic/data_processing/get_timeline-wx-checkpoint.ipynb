```python
from tqdm import tqdm

for _,rows in tqdm(a.iterrows(), total=len(a)):
    try:
        ##################正向  bus 解析##############################
        url=rows['new_url']

        headers = {"User-Agent": random.choice(ua_list)}

        response_bus_1=requests.get(url,headers=headers)
        response_bus_1.encoding='utf-8'

        soup_bus_1=BeautifulSoup(response_bus_1.text,'html.parser')



        #标题
        title_element =soup_bus_1.find("div", class_="line-title").find("h1")
        title = title_element.get_text(strip=True)
        print(title)



        #反向路径link
        alternative_sections =soup_bus_1.find_all("div", class_="alternative-directions-list")
        alternative_links = []
        for section in alternative_sections:
            for li in section.find_all("li"):
                a_tag = li.find("a", href=True)
                if a_tag:
                    link = a_tag["href"]
                    alternative_links.append(link)

   
        #提取时刻表表头
        head = [th.get_text(strip=True) for th in soup_bus_1.find("thead").find_all("th")]



    
        table_body = soup_bus_1.find("tbody", class_="stations-table-column")
        '''
        <tbody class="stations-table-column"><tr><td>港城路地铁站</td><td>04:40</td><td>04:45</td><td>04:50</td><td>04:55</td><td>05:00</td><td>05:05</td><td>05:10</td><td>05:15</td><td>05:20</td><td>05:25</td><td>05:30</td><td>05:35</td><td>05:40</td><td>05:45</td><td>05:50</td><td>05:55</td><td>06:00</td><td>06:05</td><td>06:10
        '''
 
        rows = []
        # 创建一个空列表 rows，用于保存所有行的数据。
 
        for tr in table_body.find_all("tr"):
            # 在表格主体 (table_body) 中查找所有行 (<tr>) 标签，并循环处理每一行
            #一个tr对应的是一个公交站的所有到达时间        
 
            tds = []
            for td in tr.find_all("td"):
                #在当前行 tr 中，查找所有列 (<td>) 标签，并循环处理每一列。
 
                text = td.get_text(strip=True)
                #获取列中的文本内容，并用 strip=True 去掉首尾空格和换行。
 
                # 检查是否匹配 (招呼站) 或 (临时站)，只对这种情况做清理
                if re.search(r'\((招呼站|临时站)\)', text):
                    text = re.sub(r'\((招呼站|临时站)\)', '', text).strip()
 
                tds.append(text)
            rows.append(tds)

        df = pd.DataFrame(rows, columns=head)
        df.to_csv(title+'.csv', index=False, encoding="utf-8-sig")
    except Exception as e:
        print('这个有问题',url)
        print(e)
        print(headers)

    
    ##############################################################
    time.sleep(10)



    
    ###############反向#################################
    
    for url_reverse in alternative_links:
        try:
            headers = {"User-Agent": random.choice(ua_list)}

            response_bus_1=requests.get(url_reverse,headers=headers)
            response_bus_1.encoding='utf-8'

            soup_bus_1=BeautifulSoup(response_bus_1.text,'html.parser')



            #标题
            title_element =soup_bus_1.find("div", class_="line-title").find("h1")
            title = title_element.get_text(strip=True)
            print(title)



            #提取时刻表表头
            head = [th.get_text(strip=True) for th in soup_bus_1.find("thead").find_all("th")]



            #提取时刻表
            table_body = soup_bus_1.find("tbody", class_="stations-table-column")
            '''
            <tbody class="stations-table-column"><tr><td>港城路地铁站</td><td>04:40</td><td>04:45</td><td>04:50</td><td>04:55</td><td>05:00</td><td>05:05</td><td>05:10</td><td>05:15</td><td>05:20</td><td>05:25</td><td>05:30</td><td>05:35</td><td>05:40</td><td>05:45</td><td>05:50</td><td>05:55</td><td>06:00</td><td>06:05</td><td>06:10
            '''
 
            rows = []
            # 创建一个空列表 rows，用于保存所有行的数据。
 
            for tr in table_body.find_all("tr"):
                # 在表格主体 (table_body) 中查找所有行 (<tr>) 标签，并循环处理每一行
                #一个tr对应的是一个公交站的所有到达时间        
 
                tds = []
                for td in tr.find_all("td"):
                    #在当前行 tr 中，查找所有列 (<td>) 标签，并循环处理每一列。
 
                    text = td.get_text(strip=True)
                    #获取列中的文本内容，并用 strip=True 去掉首尾空格和换行。
 
                    # 检查是否匹配 (招呼站) 或 (临时站)，只对这种情况做清理
                    if re.search(r'\((招呼站|临时站)\)', text):
                        text = re.sub(r'\((招呼站|临时站)\)', '', text).strip()
 
                    tds.append(text)
                rows.append(tds)

            df = pd.DataFrame(rows, columns=head)
            df.to_csv(title+'.csv', index=False, encoding="utf-8-sig")
        except  Exception as e:
            print('这个有问题',url_reverse)
            print(e)
            print(headers)

    
        time.sleep(2)
    #########################################################################
```
