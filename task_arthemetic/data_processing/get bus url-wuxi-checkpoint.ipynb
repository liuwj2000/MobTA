```python
import time
import logging
from selenium import webdriver
#Selenium 是用于自动化网页操作的库，这里引入了其浏览器驱动模块
from selenium.webdriver.chrome.options import Options as ChromeOptions
#给 Chrome 浏览器配置启动参数
from selenium.webdriver.common.by import By
#By: 用于元素定位


def initialize_driver():
    """初始化 Chrome 驱动"""
    option = ChromeOptions()
    option.add_argument('--ignore-certificate-errors')
    #忽略 HTTPS 证书错误
    option.add_argument('--ignore-ssl-errors')
    #忽略 SSL 错误，进一步确保不会因为证书或加密问题阻止网页加载。
    option.add_argument('--mute-audio')
    #静音浏览器，避免有声音打扰（例如自动播放视频的网页）。
    option.add_argument('--log-level=2')
    driver = webdriver.Chrome(options=option)
    driver.maximize_window()
    #最大化浏览器窗口，便于元素完整显示，减少页面元素定位失败的问题
    return driver
```

```python
def crawl_lines(driver,tag,num_limit):
    """向下滚动，获取全部公交线路"""
    scroll_viewport_selector = '.cdk-virtual-scroll-viewport'
    # 定位页面中的“可滚动区域”的CSS选择器
    line_selector = '.line-row'
    # 定位每条公交线路的CSS选择器
    link_selector = '.line-row a'
    # 定位公交线路行里链接的CSS选择器
    viewport = driver.find_element(By.CSS_SELECTOR, scroll_viewport_selector)
    # 找到这个可滚动区域元素

    count_prev = 0
    # 用来记录上一轮公交线路数
    line_text=[]
    # 用来保存公交线路文本的列表
    while count_prev<num_limit:
        #lines = driver.find_elements(By.CSS_SELECTOR, line_selector)
        links = driver.find_elements(By.CSS_SELECTOR, link_selector)
        #print(len(lines),len(links))
         # 获取当前页面上显示的公交线路元素
        tmp=[]
        for link in links:
            try:
                parent = link.find_element(By.XPATH, "./..")  # a 标签的父节点是 .line-row
                text = parent.text.strip()
                url = link.get_attribute('href')
                if text and not text.startswith(tag):  # 跳过“北京公交”这些分类行
                    tmp.append((text, url))
            except Exception as e:
                continue  # 避免异常打断
        line_text.extend(tmp)
        # 将公交线路文本全部追加到 line_text 列表
        count_new = len(line_text)
        #print(count_new,count_prev)
        if count_new == count_prev:
            break
        count_prev = count_new
        driver.execute_script("arguments[0].scrollTop +=600", viewport)
        # 向页面底部滚动
        time.sleep(2)  # 等待加载，防止过快爬虫
    # 获取公交线路文本
    return line_text
```

```python
def get_url(url,tag,num_limit=5000):

    driver = initialize_driver()
    driver.get(url)

    lines = crawl_lines(driver,tag,num_limit)
    logging.info(f"共获取到公交数: {len(lines)}")

    driver.quit()
    return lines
```

```python


dict_all={'无锡公交':
          'https://moovitapp.com/tripplan/wuxi-5315/lines/zh-cn?customerId=4908&ref=4&query=%E6%97%A0%E9%94%A1%E5%85%AC%E4%BA%A4&af_sub8=%2Findex%2Fzh-cn%2F%25E5%2585%25AC%25E5%2585%25B1%25E4%25BA%25A4%25E9%2580%259A-lines-Wuxi-5315-1099462&af_sub9=agency_show_all_lines'}

```

```python
all_urls=set()
```

```python
dict_all.keys()
```

```python

for i in dict_all.keys():
    lines=get_url(dict_all[i],i,5000)
    for url in lines:
        all_urls.add(url)
    print(i,len(all_urls))
all_urls
```

```python
import pandas as pd
all_urls_df=pd.DataFrame(all_urls)
all_urls_df
```
