```python
import requests
import time

def get_location(station_name, city, key):
    url = 'https://restapi.amap.com/v3/geocode/geo'
    params = {
        'address': station_name,
        'city': city,
        'output': 'JSON',
        'key': key
    }
    response = requests.get(url, params=params).json()
    try:
        location = response['geocodes'][0]['location']  # lng,lat
        lng, lat = map(float, location.split(','))
        return lat, lng
    except Exception:
        print(station_name)
        return None, None

api_key = 'your key'
lat, lng = get_location('浦三路东方路', '上海', api_key)
print(lat, lng)
```

```python
lat2, lng2 = get_location('浦三路东方路公交站', '上海', api_key)
print(lat2, lng2)
```

```python
lat2, lng2 = get_location('浦三路东方路(公交站)', '上海', api_key)
print(lat2, lng2)
```
